```python
import sqlite3
import pandas as pd
import time
conn = sqlite3.connect('../data/nba_stats.db')


player_info = pd.read_sql("SELECT * FROM player_info", conn)
player_boxscores_basic = pd.read_sql("SELECT * FROM player_basic_stats", conn)

player_ids_in_db = set(player_info['PERSON_ID'].tolist())

player_ids_in_boxscores = set(player_boxscores_basic['personId'].tolist())

ids_to_load = player_ids_in_boxscores - player_ids_in_db

for p_id in ids_to_load:
    info = commonplayerinfo.CommonPlayerInfo(p_id).get_data_frames()[0]
    info = info.rename(columns = {'TEAM_ID':'teamId',
                                 'TEAM_NAME':'teamName',
                                 'TEAM_ABBREVIATION':'teamTricode'})
                                 
    info.to_sql("player_info", conn, if_exists='append', index=False)
    time.sleep(2)


```

```python
from nba_api.stats.endpoints import scoreboardv3
from datetime import datetime

def get_nba_schedule(game_date=None):
    """
    Get today's games with official NBA gameIds.
    Returns DataFrame with gameId, home_team, away_team, gameDate.
    """
    if game_date is None:
        game_date = datetime.now().strftime('%Y-%m-%d')
    
    try:
        sb = scoreboardv3.ScoreboardV3(game_date=game_date)
        games = sb.get_normalized_dict()['GameHeader']
        
        if not games:
            return pd.DataFrame()
        
        records = []
        for g in games:
            records.append({
                'gameId': str(g['gameId']),
                'gameDate': game_date,
                'home_team': g['homeTeamName'],  # Check actual key names
                'away_team': g['awayTeamName'],
                'gameStatus': g.get('gameStatus', 1)
            })
        
        return pd.DataFrame(records)
    
    except Exception as e:
        return pd.DataFrame()
```
